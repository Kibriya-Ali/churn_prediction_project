```python
import pandas as pd
import numpy as np
```

```python
logs_path = r'D:\Databases\churn-prediction-project\data\raw\user_logs.csv'
chunk_size = 5000000
aggregated_chunks = []
```

```python
for chunk in pd.read_csv(logs_path, chunksize = chunk_size):
    user_summary = chunk.groupby('msno').agg({
        'num_25':'sum',
        'num_100':'sum',
        'total_secs':'sum',
        'date':'max'
    }).reset_index()

aggregated_chunks.append(user_summary)
```

```python
combined_logs = pd.concat(aggregated_chunks)
```

```python
final_user_logs = combined_logs.groupby('msno').agg({
        'num_25':'sum',
        'num_100':'sum',
        'total_secs':'sum',
        'date':'max'
    }).reset_index()
```

```python
final_user_logs.rename(columns = {
        'num_25':'total_songs_skipped',
        'num_100':'total_songs_finished',
        'total_secs':'total_time_listened_in_secs',
        'date':'last_active_date'
},inplace = True)
```

```python
final_user_logs.to_csv(r'D:\Databases\churn-prediction-project\data\processed\aggregated_logs.csv', index = False)
```
